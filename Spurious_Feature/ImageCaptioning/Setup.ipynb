### LAION-5B-10k Image Set

- This is a random 10k subset of LAION-5B images, aimed at testing model speed.

```python
import os
import shutil
import random
from tqdm import tqdm
from PIL import Image, UnidentifiedImageError

def copy_limited_files_streaming(
    source_dir,
    dest_dir,
    limit=100,
    acceptance_prob=0.01,   # tune if images are sparse
    seed=42
):
    """
    Streams through a very large directory and randomly copies up to `limit`
    valid images without listing all files into memory.

    - Uses os.scandir() (fast, lazy)
    - Stops early when limit is reached
    - Skips corrupted / invalid images
    """

    random.seed(seed)
    os.makedirs(dest_dir, exist_ok=True)

    copied = 0
    skipped = 0
    scanned = 0

    with tqdm(total=limit, desc="Copying valid images", unit="file") as pbar:
        with os.scandir(source_dir) as it:
            for entry in it:
                if copied >= limit:
                    break

                if not entry.is_file():
                    continue

                scanned += 1

                # Random down-sampling to avoid decoding everything
                if random.random() > acceptance_prob:
                    continue

                src_path = entry.path
                dest_path = os.path.join(dest_dir, entry.name)

                try:
                    with Image.open(src_path) as img:
                        w, h = img.size
                        if w <= 1 or h <= 1:
                            skipped += 1
                            continue

                    shutil.copy2(src_path, dest_path)
                    copied += 1
                    pbar.update(1)

                except (UnidentifiedImageError, OSError):
                    skipped += 1
                    continue

    print(f"\nScanned: {scanned:,}")
    print(f"Copied: {copied:,}")
    print(f"Skipped: {skipped:,}")

# Example usage
source = r"G:\Thesis\0000_images"
dest = r"EvaluationDatasets\LAION-5B-10k\LAION-5B-10k-images"
limit = 10_000

copy_limited_files_streaming(source, dest, limit)
```
