# Facial Detection

- This notebook contains code to carry out facial detection & cropping required for gender & skin-tone labelling.

Metrics:
- Precision and Recall at different IoU thresholds
- AP (Average Precision) across multiple IoU thresholds
- Detection rate by face size: Small (<32px), Medium (32-96px), Large (>96px)
- False Positive Rate
- Inference time (FPS)

-> https://claude.ai/chat/45762296-64fc-4701-8840-f81a3eb24466

```python
import os
import cv2
import numpy as np
from pathlib import Path
from typing import List, Tuple
import time
```

```python
# =============================================================================
# SHARED UTILITIES
# =============================================================================

def crop_face(image: np.ndarray, bbox: Tuple[int, int, int, int], padding: int = 20) -> np.ndarray:
    """Crop face from image with padding"""
    x1, y1, x2, y2 = bbox
    h, w = image.shape[:2]
    
    # Add padding
    x1 = max(0, x1 - padding)
    y1 = max(0, y1 - padding)
    x2 = min(w, x2 + padding)
    y2 = min(h, y2 + padding)
    
    return image[y1:y2, x1:x2]

def save_cropped_faces(image: np.ndarray, boxes: List[Tuple[int, int, int, int]], 
                       image_name: str, output_dir: Path, padding: int = 20) -> int:
    """Save all detected faces from an image"""
    for face_idx, bbox in enumerate(boxes, 1):
        face_crop = crop_face(image, bbox, padding)
        output_path = output_dir / f"{image_name}-face-{face_idx}.jpg"
        cv2.imwrite(str(output_path), face_crop)
    
    return len(boxes)

def collect_image_files(input_dir: str, extensions: List[str] = ['.jpg', '.jpeg', '.png']) -> List[str]:
    """Collect all image files from directory"""
    input_path = Path(input_dir)
    image_files = []
    
    for ext in extensions:
        image_files.extend(input_path.glob(f"*{ext}"))
        # image_files.extend(input_path.glob(f"*{ext.upper()}"))
    
    return [str(f) for f in image_files]

def print_stats(detector_name: str, stats: dict) -> None:
    """Print processing statistics"""
    print("\n" + "="*50)
    print(f"Detector: {detector_name}")
    print(f"Total images processed: {stats['total_images']}")
    print(f"Images with faces: {stats['images_with_faces']}")
    print(f"Total faces detected: {stats['total_faces']}")
    print(f"Average faces per image: {stats['total_faces'] / max(1, stats['total_images']):.2f}")
    print(f"Processing time: {stats['processing_time']:.2f}s")
    print(f"Average time per image: {stats['processing_time'] / max(1, stats['total_images']):.3f}s")
    print("="*50 + "\n")
```

## Retina Face

Link: https://github.com/serengil/retinaface

```python
import os
import cv2
import numpy as np
from pathlib import Path
from typing import List, Tuple
import time

def process_with_retinaface(input_dir: str, output_dir: str, batch_size: int = 8, confidence_threshold: float = 0.5, padding: int = 20):
    """Process images with RetinaFace detector"""
    try:
        from retinaface import RetinaFace
    except ImportError:
        print("RetinaFace not installed. Run: pip install retinaface-pytorch")
        return None
    
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)
    
    image_files = collect_image_files(input_dir)
    total_images = len(image_files)
    
    if total_images == 0:
        print(f"No images found in {input_dir}")
        return None
    
    stats = {
        'total_images': 0,
        'total_faces': 0,
        'images_with_faces': 0,
        'processing_time': 0
    }
    
    print(f"Processing {total_images} images with RetinaFace...")
    
    for i in range(0, total_images, batch_size):
        batch_files = image_files[i:i + batch_size]
        
        for img_path in batch_files:
            image = cv2.imread(img_path)
            if image is None:
                print(f"Warning: Could not read {img_path}")
                continue
            
            # RetinaFace expects RGB
            rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
            
            start_time = time.time()
            faces = RetinaFace.detect_faces(rgb_image)
            stats['processing_time'] += time.time() - start_time
            
            boxes = []
            if isinstance(faces, dict):
                for key, face_data in faces.items():
                    if face_data['score'] >= confidence_threshold:
                        box = face_data['facial_area']
                        boxes.append((box[0], box[1], box[2], box[3]))
            
            image_name = Path(img_path).stem
            face_count = save_cropped_faces(image, boxes, image_name, output_path, padding)
            
            stats['total_images'] += 1
            stats['total_faces'] += face_count
            if face_count > 0:
                stats['images_with_faces'] += 1
        
        processed = min(i + batch_size, total_images)
        print(f"Progress: {processed}/{total_images} images processed", end='\r')
    
    print()
    print_stats("RetinaFace", stats)
    return stats
```

## SCRFD

Link: https://github.com/deepinsight/insightface/tree/master

```python
import os
import cv2
import numpy as np
from pathlib import Path
from typing import List, Tuple
import time

def process_with_scrfd(input_dir: str, output_dir: str, batch_size: int = 8, confidence_threshold: float = 0.5, padding: int = 20):
    """Process images with SCRFD detector from InsightFace"""
    try:
        from insightface.app import FaceAnalysis
    except ImportError:
        print("InsightFace not installed. Run: pip install insightface onnxruntime")
        return None
    
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)
    
    # Initialize detector
    app = FaceAnalysis(allowed_modules=['detection'])
    app.prepare(ctx_id=0, det_thresh=confidence_threshold)
    
    image_files = collect_image_files(input_dir)
    total_images = len(image_files)
    
    if total_images == 0:
        print(f"No images found in {input_dir}")
        return None
    
    stats = {
        'total_images': 0,
        'total_faces': 0,
        'images_with_faces': 0,
        'processing_time': 0
    }
    
    print(f"Processing {total_images} images with SCRFD...")
    
    for i in range(0, total_images, batch_size):
        batch_files = image_files[i:i + batch_size]
        
        for img_path in batch_files:
            image = cv2.imread(img_path)
            if image is None:
                print(f"Warning: Could not read {img_path}")
                continue
            
            start_time = time.time()
            faces = app.get(image)
            stats['processing_time'] += time.time() - start_time
            
            boxes = []
            for face in faces:
                bbox = face.bbox.astype(int)
                boxes.append((bbox[0], bbox[1], bbox[2], bbox[3]))
            
            image_name = Path(img_path).stem
            face_count = save_cropped_faces(image, boxes, image_name, output_path, padding)
            
            stats['total_images'] += 1
            stats['total_faces'] += face_count
            if face_count > 0:
                stats['images_with_faces'] += 1
        
        processed = min(i + batch_size, total_images)
        print(f"Progress: {processed}/{total_images} images processed", end='\r')
    
    print()
    print_stats("SCRFD", stats)
    return stats
```

## YOLO-face

Link: https://github.com/YapaLab/yolo-face

```python
import os
import cv2
import numpy as np
from pathlib import Path
from typing import List, Tuple
import time

def process_with_yolo_face(input_dir: str, output_dir: str, model_path: str, batch_size: int = 8, confidence_threshold: float = 0.5, padding: int = 20):
    """Process images with YOLOv8-Face detector"""
    try:
        from ultralytics import YOLO
    except ImportError:
        print("Ultralytics not installed. Run: pip install ultralytics")
        return None
    
    if not os.path.exists(model_path):
        print(f"Model not found: {model_path}")
        return None
    
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)
    
    # Load model
    model = YOLO(model_path)
    
    image_files = collect_image_files(input_dir)
    total_images = len(image_files)
    
    if total_images == 0:
        print(f"No images found in {input_dir}")
        return None
    
    stats = {
        'total_images': 0,
        'total_faces': 0,
        'images_with_faces': 0,
        'processing_time': 0
    }
    
    print(f"Processing {total_images} images with YOLOv8-Face...")
    
    for i in range(0, total_images, batch_size):
        batch_files = image_files[i:i + batch_size]
        
        # Read batch images
        batch_images = []
        valid_files = []
        for img_path in batch_files:
            img = cv2.imread(img_path)
            if img is not None:
                batch_images.append(img)
                valid_files.append(img_path)
        
        if not batch_images:
            continue
        
        # Process batch
        start_time = time.time()
        results = model(batch_images, conf=confidence_threshold, verbose=False)
        stats['processing_time'] += time.time() - start_time
        
        # Save results
        for img, result, img_path in zip(batch_images, results, valid_files):
            boxes = []
            if result.boxes is not None:
                for box in result.boxes:
                    x1, y1, x2, y2 = box.xyxy[0].cpu().numpy()
                    boxes.append((int(x1), int(y1), int(x2), int(y2)))
            
            image_name = Path(img_path).stem
            face_count = save_cropped_faces(img, boxes, image_name, output_path, padding)
            
            stats['total_images'] += 1
            stats['total_faces'] += face_count
            if face_count > 0:
                stats['images_with_faces'] += 1
        
        processed = min(i + batch_size, total_images)
        print(f"Progress: {processed}/{total_images} images processed", end='\r')
    
    print()
    print_stats("YOLO-Face", stats)
    return stats
```

## MediaPipe Face Detection

Link: https://mediapipe.readthedocs.io/en/latest/solutions/face_detection.html

```python
import os
import cv2
import numpy as np
from pathlib import Path
from typing import List, Tuple
import time

def process_with_mediapipe(input_dir: str, output_dir: str, batch_size: int = 8, confidence_threshold: float = 0.5, padding: int = 20):
    """Process images with MediaPipe Face Detection"""
    try:
        import mediapipe as mp
    except ImportError:
        print("MediaPipe not installed. Run: pip install mediapipe")
        return None
    
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)
    
    # Initialize detector
    mp_face_detection = mp.solutions.face_detection
    detector = mp_face_detection.FaceDetection(
        model_selection=1,  # 1 for full range, 0 for short range
        min_detection_confidence=confidence_threshold
    )
    
    image_files = collect_image_files(input_dir)
    total_images = len(image_files)
    
    if total_images == 0:
        print(f"No images found in {input_dir}")
        return None
    
    stats = {
        'total_images': 0,
        'total_faces': 0,
        'images_with_faces': 0,
        'processing_time': 0
    }
    
    print(f"Processing {total_images} images with MediaPipe...")
    
    for i in range(0, total_images, batch_size):
        batch_files = image_files[i:i + batch_size]
        
        for img_path in batch_files:
            image = cv2.imread(img_path)
            if image is None:
                print(f"Warning: Could not read {img_path}")
                continue
            
            # MediaPipe expects RGB
            rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
            
            start_time = time.time()
            results = detector.process(rgb_image)
            stats['processing_time'] += time.time() - start_time
            
            boxes = []
            if results.detections:
                h, w, _ = image.shape
                for detection in results.detections:
                    bbox = detection.location_data.relative_bounding_box
                    x1 = int(bbox.xmin * w)
                    y1 = int(bbox.ymin * h)
                    x2 = int((bbox.xmin + bbox.width) * w)
                    y2 = int((bbox.ymin + bbox.height) * h)
                    boxes.append((x1, y1, x2, y2))
            
            image_name = Path(img_path).stem
            face_count = save_cropped_faces(image, boxes, image_name, output_path, padding)
            
            stats['total_images'] += 1
            stats['total_faces'] += face_count
            if face_count > 0:
                stats['images_with_faces'] += 1
        
        processed = min(i + batch_size, total_images)
        print(f"Progress: {processed}/{total_images} images processed", end='\r')
    
    print()
    print_stats("MediaPipe", stats)
    return stats
```

```python
def plot_faces_on_images(dataset, preds, save_dir):
    """
    Draw predicted bounding boxes on images and save them.

    Args:
        dataset: Hugging Face dataset containing 'image' field.
        preds: List of prediction dicts with 'boxes' (and optionally 'filename').
        save_dir: Directory to save annotated images.
    """
    save_path = Path(save_dir)
    save_path.mkdir(parents=True, exist_ok=True)

    for sample, pred in zip(dataset, preds):
        image = np.array(sample['image'].convert('RGB')).copy()
        for box in pred['boxes']:
            x, y, w, h = box
            cv2.rectangle(image, (x, y), (x + w, y + h), color=(0, 255, 0), thickness=2)

        filename = pred.get('filename', getattr(sample['image'], 'filename', 'image.png'))
        save_file = save_path / Path(filename).name
        cv2.imwrite(str(save_file), cv2.cvtColor(image, cv2.COLOR_RGB2BGR))
```

## Retina Face

- Batch Face Implements batching and GPU capabilities for Retina Face Model: https://pypi.org/project/batch-face/

```python
from batch_face import RetinaFace
from tqdm import tqdm
import numpy as np
import time

def evaluate_retina_face(dataset, confidence_threshold=0.5, batch_size=32, gpu_id=0, fp16=True, save_filenames=False):
    """
    Evaluate RetinaFace using batch_face on GPU with batch input support.

    Args:
        dataset: Hugging Face dataset containing 'image' and 'faces' fields.
        confidence_threshold: Minimum score for a detection to be considered.
        batch_size: Number of images to process in a single batch.
        gpu_id: GPU index to use. Set -1 for CPU.
        fp16: Whether to use FP16 for faster inference on GPU.
        save_filenames: If True, stores image filenames in preds for later plotting.

    Returns:
        preds: List of dicts {'boxes': [...], 'scores': [...], 'image_name': str (optional)}
        gts: List of ground-truth bounding boxes
        inference_times: List of per-image inference times
    """
    # Initialize detector
    detector = RetinaFace(gpu_id=gpu_id, fp16=fp16)

    # Convert dataset to list of images and optional filenames
    images = [np.array(sample['image'].convert('RGB')) for sample in dataset]
    gts = [sample['faces']['bbox'] for sample in dataset]
    image_names = [sample['image'].filename if save_filenames else None for sample in dataset]

    preds = []
    inference_times = []

    for i in tqdm(range(0, len(images), batch_size), desc="Evaluating RetinaFace (GPU)", unit="batch"):
        batch_imgs = images[i:i+batch_size]
        batch_names = image_names[i:i+batch_size] if save_filenames else [None]*len(batch_imgs)

        start = time.time()
        all_faces = detector(batch_imgs, batch_size=len(batch_imgs), return_dict=True)
        batch_time = time.time() - start

        per_image_time = batch_time / len(batch_imgs)
        inference_times.extend([per_image_time] * len(batch_imgs))

        # Collect predictions
        for faces, img_name in zip(all_faces, batch_names):
            boxes, scores = [], []
            for face in faces:
                if face['score'] >= confidence_threshold:
                    x1, y1, x2, y2 = face['box']
                    boxes.append([int(x1), int(y1), int(x2 - x1), int(y2 - y1)])
                    scores.append(face['score'])
            pred_dict = {'boxes': boxes, 'scores': scores}
            if save_filenames:
                pred_dict['image_name'] = img_name
            preds.append(pred_dict)

    return preds, gts, inference_times
```

```python
from datasets import load_dataset

wider_face_dataset = load_dataset("CUHK-CSE/wider_face", split="validation", trust_remote_code=True)
retina_face_preds, retina_face_gts, retina_face_times = evaluate_retina_face(wider_face_dataset, confidence_threshold=0.5, batch_size=32, save_filenames=True)
# Directory where you want to save annotated images
save_dir = "annotated_images/retina_face"
plot_faces_on_images(wider_face_dataset, retina_face_preds, save_dir)
print(f"Annotated images saved to: {save_dir}")
```

## SCRFD

```python
from insightface.app import FaceAnalysis
from tqdm import tqdm
import numpy as np
import time

def evaluate_scrfd(dataset, confidence_threshold=0.5, batch_size=32, ctx_id=0, save_filenames=False):
    """
    Evaluate SCRFD detector from InsightFace with batch processing on GPU.

    Args:
        dataset: Hugging Face dataset containing 'image' and 'faces' fields.
        confidence_threshold: Minimum score for a detection to be considered.
        batch_size: Number of images to process in a single batch.
        ctx_id: GPU index (0,1,...) or -1 for CPU.
        save_filenames: If True, stores image filenames in preds for later plotting.

    Returns:
        preds: List of dicts {'boxes': [...], 'scores': [...], 'image_name': str (optional)}
        gts: List of ground-truth bounding boxes
        inference_times: List of per-image inference times
    """
    # Initialize SCRFD detector
    app = FaceAnalysis(allowed_modules=['detection'])
    app.prepare(ctx_id=ctx_id, det_thresh=confidence_threshold)

    # Convert dataset to list of images and optional filenames
    images = [np.array(sample['image'].convert('RGB')) for sample in dataset]
    gts = [sample['faces']['bbox'] for sample in dataset]
    image_names = [sample['image'].filename if save_filenames else None for sample in dataset]

    preds = []
    inference_times = []

    for i in tqdm(range(0, len(images), batch_size), desc="Evaluating SCRFD", unit="batch"):
        batch_imgs = images[i:i+batch_size]
        batch_names = image_names[i:i+batch_size] if save_filenames else [None]*len(batch_imgs)

        start = time.time()
        # Batch processing: SCRFD doesn't natively support batch, so we still loop here
        batch_faces = [app.get(img) for img in batch_imgs]
        batch_time = time.time() - start

        per_image_time = batch_time / len(batch_imgs)
        inference_times.extend([per_image_time] * len(batch_imgs))

        for faces, img_name in zip(batch_faces, batch_names):
            boxes, scores = [], []
            for face in faces:
                if face.det_score >= confidence_threshold:
                    x1, y1, x2, y2 = face.bbox.astype(int)
                    boxes.append([int(x1), int(y1), int(x2 - x1), int(y2 - y1)])
                    scores.append(face.det_score)
            pred_dict = {'boxes': boxes, 'scores': scores}
            if save_filenames:
                pred_dict['image_name'] = img_name
            preds.append(pred_dict)

    return preds, gts, inference_times
```

## YOLO-Face

```python
from ultralytics import YOLO
from tqdm import tqdm
import numpy as np
import time

def evaluate_yolo_face(dataset, model_path, confidence_threshold=0.5, batch_size=16, device=0, save_filenames=False):
    """
    Evaluate YOLOv8-Face detector on a dataset with batch input and GPU support.

    Args:
        dataset: Hugging Face dataset containing 'image' and 'faces' fields.
        model_path: Path to the trained YOLOv8-Face model (.pt or .yaml).
        confidence_threshold: Minimum score for a detection to be considered.
        batch_size: Number of images to process in a single batch.
        device: GPU index (0,1,...) or 'cpu' for CPU inference.
        save_filenames: If True, saves the filename with each prediction.

    Returns:
        preds: List of dicts {'boxes': [...], 'scores': [...], 'filename': ... (optional)}
        gts: List of ground-truth bounding boxes
        inference_times: List of per-image inference times
    """
    # Load YOLO model
    model = YOLO(model_path)
    model.fuse()  # optional: fuse Conv+BN for faster inference
    model.to(device)

    # Prepare images, filenames, and ground truth
    images = [np.array(sample['image'].convert('RGB')) for sample in dataset]
    filenames = [sample['image'].filename if hasattr(sample['image'], 'filename') else f'image_{i}' 
                 for i, sample in enumerate(dataset)]
    gts = [sample['faces']['bbox'] for sample in dataset]

    preds = []
    inference_times = []

    for i in tqdm(range(0, len(images), batch_size), desc="Evaluating YOLO-Face", unit="batch"):
        batch_imgs = images[i:i+batch_size]
        batch_files = filenames[i:i+batch_size]

        start = time.time()
        results = model(batch_imgs, conf=confidence_threshold, verbose=False)
        batch_time = time.time() - start
        per_image_time = batch_time / len(batch_imgs)
        inference_times.extend([per_image_time] * len(batch_imgs))

        for result, fname in zip(results, batch_files):
            boxes, scores = [], []
            if result.boxes is not None:
                for box in result.boxes:
                    x1, y1, x2, y2 = box.xyxy[0].cpu().numpy()
                    boxes.append([int(x1), int(y1), int(x2 - x1), int(y2 - y1)])  # [x, y, w, h]
                    scores.append(float(box.conf.cpu()))
            pred_dict = {'boxes': boxes, 'scores': scores}
            if save_filenames:
                pred_dict['filename'] = Path(fname).name
            preds.append(pred_dict)

    return preds, gts, inference_times
```

```python
from datasets import load_dataset

wider_face_dataset = load_dataset("CUHK-CSE/wider_face", split="validation", trust_remote_code=True)
yolo_preds, yolo_gts, yolo_times = evaluate_yolo_face(wider_face_dataset, model_path="yolov12l-face.pt", confidence_threshold=0.5, batch_size=32, device=0, save_filenames=True)
# Directory where you want to save annotated images
save_dir = "annotated_images/yolo"
plot_faces_on_images(wider_face_dataset, yolo_preds, save_dir)
print(f"Annotated images saved to: {save_dir}")
```

## MediaPipe Face Detection

- Does not support Batching

```python
import mediapipe as mp
from tqdm import tqdm
import numpy as np
import time

def evaluate_mediapipe(dataset, confidence_threshold=0.5, batch_size=16, save_filenames=False):
    """
    Evaluate MediaPipe Face Detection on a dataset with pseudo-batching.
    
    Args:
        dataset: Hugging Face dataset containing 'image' and 'faces' fields.
        confidence_threshold: Minimum detection score to consider a face.
        batch_size: Number of images to process per loop iteration.
        save_filenames: If True, stores image filenames in preds for later plotting.
    
    Returns:
        preds: List of dicts {'boxes': [...], 'scores': [...], 'image_name': str (optional)}
        gts: List of ground-truth bounding boxes
        inference_times: List of per-image inference times
    """
    mp_face_detection = mp.solutions.face_detection
    detector = mp_face_detection.FaceDetection(
        model_selection=1,  # 1 for full range, 0 for short range
        min_detection_confidence=confidence_threshold
    )

    # Prepare images and optional filenames
    images = [np.array(sample['image'].convert('RGB')) for sample in dataset]
    gts = [sample['faces']['bbox'] for sample in dataset]
    image_names = [sample['image'].filename if save_filenames else None for sample in dataset]

    preds = []
    inference_times = []

    for i in tqdm(range(0, len(images), batch_size), desc="Evaluating MediaPipe", unit="batch"):
        batch_imgs = images[i:i+batch_size]
        batch_names = image_names[i:i+batch_size] if save_filenames else [None]*len(batch_imgs)

        for img, img_name in zip(batch_imgs, batch_names):
            start = time.time()
            results = detector.process(img)
            elapsed = time.time() - start
            inference_times.append(elapsed)

            boxes, scores = [], []
            if results.detections:
                h, w, _ = img.shape
                for det in results.detections:
                    bbox = det.location_data.relative_bounding_box
                    x1 = int(bbox.xmin * w)
                    y1 = int(bbox.ymin * h)
                    x2 = int((bbox.xmin + bbox.width) * w)
                    y2 = int((bbox.ymin + bbox.height) * h)
                    boxes.append([int(x1), int(y1), int(x2 - x1), int(y2 - y1)])
                    score = float(det.score[0]) if det.score else 1.0
                    scores.append(score)

            pred_dict = {'boxes': boxes, 'scores': scores}
            if save_filenames:
                pred_dict['image_name'] = img_name
            preds.append(pred_dict)

    return preds, gts, inference_times
```

```python
from datasets import load_dataset

wider_face_dataset = load_dataset("CUHK-CSE/wider_face", split="validation", trust_remote_code=True)
mediapipe_preds, mediapipe_gts, mediapipe_times = evaluate_mediapipe(wider_face_dataset, confidence_threshold=0.5, batch_size=32, save_filenames=True)
# Directory where you want to save annotated images
save_dir = "annotated_images/mediapipe"
plot_faces_on_images(wider_face_dataset, mediapipe_preds, save_dir)
print(f"Annotated images saved to: {save_dir}")
```

```python
from pycocotools.coco import COCO
from pycocotools.cocoeval import COCOeval
from pathlib import Path
import json
import numpy as np

# These need to be changed as currently they don't use the actual image size instead opting fro width=1 & depth=1 plus their are other official approaches for testing on wider Faces datasets
def prepare_coco_annotations(preds, gts, image_ids=None):
    """
    Convert your preds and gts into COCO JSON format (with required 'info' and 'licenses').
    """
    coco_gt = {
        'info': {'description': 'Face Dataset', 'version': '1.0', 'year': 2025},
        'licenses': [{'id': 1, 'name': 'Unknown', 'url': ''}],
        'images': [],
        'annotations': [],
        'categories': [{'id': 1, 'name': 'face'}]
    }
    coco_dt = []

    if image_ids is None:
        image_ids = list(range(len(preds)))

    ann_id = 1
    for img_id, gt_boxes in zip(image_ids, gts):
        coco_gt['images'].append({'id': img_id, 'width': 1, 'height': 1})
        for box in gt_boxes:
            # x, y, w, h = box
            x, y, w, h = map(int, box)  # convert to Python int
            coco_gt['annotations'].append({
                'id': ann_id,
                'image_id': img_id,
                'category_id': 1,
                'bbox': [x, y, w, h],
                'area': w * h,
                'iscrowd': 0
            })
            ann_id += 1

    for img_id, pred in zip(image_ids, preds):
        for box, score in zip(pred['boxes'], pred['scores']):
            # x, y, w, h = box
            x, y, w, h = map(int, box)          # convert coordinates to int
            score = float(score)                # convert score to float
            coco_dt.append({
                'image_id': img_id,
                'category_id': 1,
                'bbox': [x, y, w, h],
                'score': score
            })

    return coco_gt, coco_dt

def evaluate_yolo_face_coco(preds, gts):
    """
    Compute COCO metrics for YOLO-Face predictions.
    """
    coco_gt_dict, coco_dt_list = prepare_coco_annotations(preds, gts)

    # Save temporary JSON files
    tmp_gt_file = Path("tmp_gt.json")
    tmp_dt_file = Path("tmp_dt.json")
    with open(tmp_gt_file, 'w') as f:
        json.dump(coco_gt_dict, f)
    with open(tmp_dt_file, 'w') as f:
        json.dump(coco_dt_list, f)

    # Load COCO API
    coco_gt = COCO(str(tmp_gt_file))
    coco_dt = coco_gt.loadRes(str(tmp_dt_file))

    coco_eval = COCOeval(coco_gt, coco_dt, iouType='bbox')
    coco_eval.evaluate()
    coco_eval.accumulate()
    coco_eval.summarize()  # prints mAP, AP50, AP75, etc.

    # Clean up temp files
    tmp_gt_file.unlink()
    tmp_dt_file.unlink()
    
    return coco_eval
```
