```python
import torch

print("Torch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("CUDA runtime version:", torch.version.cuda)
print("GPU count:", torch.cuda.device_count())

if torch.cuda.is_available():
    print("GPU name:", torch.cuda.get_device_name(0))

    # Simple GPU compute test
    x = torch.randn(4096, 4096, device="cuda")
    y = torch.matmul(x, x)
    print("GPU computation successful:", y.mean().item())
else:
    print("CUDA is NOT working")
```

```python
from safetensors.torch import load_file
import torch

# Load the weights
weights = load_file("FairGen/output/gender/female/gender_female_1000steps.safetensors")

print("Total parameters:", len(weights))
print("\nWeight statistics:")
for k, v in weights.items():
    if 'lora' in k:
        print(f"{k}:")
        print(f"  shape: {v.shape}")
        print(f"  norm: {v.norm().item():.6f}")
        print(f"  mean: {v.mean().item():.6f}")
        print(f"  std: {v.std().item():.6f}")
        print(f"  min: {v.min().item():.6f}")
        print(f"  max: {v.max().item():.6f}")
        print()
```
