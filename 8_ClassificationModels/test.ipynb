```python
import json
from pathlib import Path
from collections import Counter


def inspect_labels(json_path: Path):

    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    print("=" * 70)
    print(f"Inspecting: {json_path}")
    print("=" * 70)

    for split in ["train", "val", "test"]:

        if split not in data:
            print(f"\n[WARNING] Missing split: {split}")
            continue

        labels = [item["label"] for item in data[split]]
        counter = Counter(labels)

        unique_labels = sorted(counter.keys())

        print(f"\n--- {split.upper()} ---")
        print(f"Samples: {len(labels)}")
        print(f"Unique labels: {unique_labels}")
        print(f"Number of classes: {len(unique_labels)}")

        if labels:
            print(f"Min label: {min(labels)}")
            print(f"Max label: {max(labels)}")

        print("\nLabel counts:")
        for label in unique_labels:
            print(f"  label {label}: {counter[label]}")

    print("\n" + "=" * 70)


if __name__ == "__main__":

    json_file = Path(
        r"F:\ImageRetrieval\SpuriousFeatureImages\StableDiffusionImages\_SPLITS\Base\splits_10mst_face_stratified.json"
    )

    inspect_labels(json_file)
```
