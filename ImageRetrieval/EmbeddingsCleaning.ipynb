```python
import os
import glob
import pyarrow.parquet as pq

def count_non_null_rows_in_parquets(folder_path, column="embeddings_result", pattern="*.parquet"):
    """
    Efficiently count non-null rows in a specific column across all Parquet files,
    reading one row group at a time to minimize memory usage.
    """
    parquet_files = glob.glob(os.path.join(folder_path, pattern))
    print(f"🗂️ Found {len(parquet_files)} parquet files")

    total_non_null = 0
    total_rows = 0

    for i, file_path in enumerate(parquet_files, 1):
        try:
            parquet_file = pq.ParquetFile(file_path)
            n_row_groups = parquet_file.num_row_groups

            print(f"[{i}/{len(parquet_files)}] Processing {os.path.basename(file_path)} ({n_row_groups} row groups)")

            for rg_idx in range(n_row_groups):
                # Read only the specified column from this row group
                table = parquet_file.read_row_group(rg_idx, columns=[column])

                # Get Arrow column (ChunkedArray)
                col_data = table[column]

                # Count non-nulls efficiently
                non_null_count = len(col_data) - col_data.null_count

                total_non_null += non_null_count
                total_rows += len(col_data)

        except Exception as e:
            print(f"⚠️ Failed to process {file_path}: {e}")

    print(f"\n✅ Finished scanning all Parquet files")
    print(f"📊 Total rows: {total_rows:,}")
    print(f"📈 Non-null '{column}' rows: {total_non_null:,}")
    print(f"🚫 Null '{column}' rows: {total_rows - total_non_null:,}")

    return total_non_null, total_rows
```

```python
import os

def count_images_fast(directory, extensions=None, recursive=False):
    """
    Count number of image files in a directory efficiently.
    Handles millions of files without loading them all into memory.

    Args:
        directory (str): Path to directory.
        extensions (list[str], optional): Image file extensions to count.
        recursive (bool): Whether to count images in subdirectories.

    Returns:
        int: Total number of image files found.
    """
    if extensions is None:
        extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.gif', '.webp']

    extensions = tuple(ext.lower() for ext in extensions)
    count = 0

    if recursive:
        # Efficient recursive walk
        for root, _, files in os.walk(directory):
            for f in files:
                if f.lower().endswith(extensions):
                    count += 1
    else:
        # Stream entries in a single directory (very fast)
        with os.scandir(directory) as it:
            for entry in it:
                if entry.is_file() and entry.name.lower().endswith(extensions):
                    count += 1

    return count
```

```python
import os
import pyarrow as pa
import pyarrow.parquet as pq
import pyarrow.compute as pc
from tqdm import tqdm

def clean_parquet_by_missing_images(src_folder, dst_folder, image_dir,
                                    pattern="*.parquet", index_col="image_index",
                                    embedding_col="embeddings_result"):
    """
    Scans parquet files in src_folder, sets embedding_col to null if corresponding
    image file (from image_dir) does not exist. Saves cleaned files to dst_folder.

    Args:
        src_folder (str): Folder containing original parquet files.
        dst_folder (str): Folder where cleaned parquet files will be saved.
        image_dir (str): Directory containing the image files.
        pattern (str): Parquet filename pattern.
        index_col (str): Column name containing the image filename/index (without extension).
        embedding_col (str): Column to nullify if image missing.
    """
    import glob
    os.makedirs(dst_folder, exist_ok=True)
    parquet_files = glob.glob(os.path.join(src_folder, pattern))
    print(f"🗂️ Found {len(parquet_files)} parquet files")

    # Create a fast set of image names with .jpg extension
    print(f"📸 Scanning existing images in: {image_dir}")
    image_set = set(os.listdir(image_dir))
    print(f"✅ Found {len(image_set):,} images")

    for i, file_path in enumerate(parquet_files, 1):
        file_name = os.path.basename(file_path)
        dst_path = os.path.join(dst_folder, file_name)
        print(f"\n[{i}/{len(parquet_files)}] Processing {file_name}")

        parquet_file = pq.ParquetFile(file_path)
        n_row_groups = parquet_file.num_row_groups

        writer = None  # ParquetWriter for incremental saving

        for rg_idx in tqdm(range(n_row_groups), desc=f"RowGroups in {file_name}", unit="group"):
            table = parquet_file.read_row_group(rg_idx)

            if index_col not in table.column_names or embedding_col not in table.column_names:
                print(f"⚠️ Skipping row group {rg_idx}: missing '{index_col}' or '{embedding_col}'")
                continue

            image_indices = table[index_col]
            embeddings = table[embedding_col]

            # Append '.jpg' to indices for existence check
            exists_mask = pa.array([
                f"{str(img)}.jpg" in image_set for img in image_indices.to_pylist()
            ])

            cleaned_embeddings = pc.if_else(
                exists_mask, embeddings, pa.nulls(len(embeddings), type=embeddings.type)
            )

            cleaned_table = table.set_column(
                table.column_names.index(embedding_col),
                embedding_col,
                cleaned_embeddings
            )

            if writer is None:
                writer = pq.ParquetWriter(dst_path, cleaned_table.schema)
            writer.write_table(cleaned_table)

        if writer is not None:
            writer.close()
            print(f"💾 Saved cleaned parquet to {dst_path}")
        else:
            print(f"⚠️ No valid data written for {file_name}")

    print("\n✅ All parquet files processed and cleaned.")
```

```python
import os
import shutil
import pyarrow as pa
import pyarrow.parquet as pq
from tqdm import tqdm

def move_images_without_embeddings(src_folder, image_dir, dst_dir,
                                   pattern="*.parquet", index_col="image_index",
                                   embedding_col="embeddings_result"):
    import glob
    os.makedirs(dst_dir, exist_ok=True)
    parquet_files = glob.glob(os.path.join(src_folder, pattern))
    print(f"🗂️ Found {len(parquet_files)} parquet files")

    valid_indices = set()

    for i, file_path in enumerate(parquet_files, 1):
        file_name = os.path.basename(file_path)
        print(f"\n[{i}/{len(parquet_files)}] Processing {file_name}")
        parquet_file = pq.ParquetFile(file_path)
        n_row_groups = parquet_file.num_row_groups

        for rg_idx in tqdm(range(n_row_groups), desc=f"RowGroups in {file_name}", unit="group"):
            table = parquet_file.read_row_group(rg_idx, columns=[index_col, embedding_col])

            indices = table[index_col].to_pylist()
            embeddings = table[embedding_col]

            # Convert ChunkedArray to single flat array if needed
            if isinstance(embeddings, pa.ChunkedArray):
                embeddings_flat = []
                for chunk in embeddings.chunks:
                    embeddings_flat.extend(chunk.to_pylist())
                embeddings = embeddings_flat
            else:
                embeddings = embeddings.to_pylist()

            # Collect only non-null embeddings
            for idx, emb in zip(indices, embeddings):
                if emb is not None:
                    valid_indices.add(str(idx))

    print(f"\n✅ Found {len(valid_indices):,} images with valid embeddings")

    # Move images without valid embeddings
    image_files = os.listdir(image_dir)
    moved_count = 0

    for img_name in tqdm(image_files, desc="Checking images", unit="image"):
        name_no_ext, ext = os.path.splitext(img_name)
        if name_no_ext not in valid_indices:
            src_path = os.path.join(image_dir, img_name)
            dst_path = os.path.join(dst_dir, img_name)
            shutil.move(src_path, dst_path)
            moved_count += 1

    print(f"\n✅ Moved {moved_count:,} images without embeddings to {dst_dir}")
```

```python
import os
import glob
import json
import pyarrow as pa
import pyarrow.parquet as pq
import pyarrow.compute as pc
from tqdm import tqdm

def reconcile_embeddings_and_images_safe(
        embeddings_dir,
        images_dir,
        state_file,
        index_col="image_index",
        embedding_col="embeddings_result",
        pattern="*.parquet"):

    # ------------------------------------------------------------
    # Load or initialize state
    # ------------------------------------------------------------
    if os.path.exists(state_file):
        with open(state_file, "r") as f:
            state = json.load(f)
    else:
        state = {
            "completed": {},  # {filename: [rowgroups]}
            "valid_embedding_indices": [],
            "image_indices": [],
            "finished": False
        }

    valid_embedding_indices = set(state["valid_embedding_indices"])
    image_indices = set(state["image_indices"])

    # ------------------------------------------------------------
    # Scan images once
    # ------------------------------------------------------------
    print(f"Scanning images in: {images_dir}")
    all_images = os.listdir(images_dir)
    current_image_indices = {os.path.splitext(f)[0] for f in all_images}

    image_indices.update(current_image_indices)

    # ------------------------------------------------------------
    # Get Parquet files
    # ------------------------------------------------------------
    parquet_files = sorted(glob.glob(os.path.join(embeddings_dir, pattern)))
    print(f"Found {len(parquet_files)} embedding files")

    # ------------------------------------------------------------
    # Overall progress bar (file-level)
    # ------------------------------------------------------------
    for file_path in tqdm(parquet_files, desc="Overall Progress", unit="file"):

        file_name = os.path.basename(file_path)
        pf = pq.ParquetFile(file_path)
        n_row_groups = pf.num_row_groups

        completed_groups = set(state["completed"].get(file_name, []))

        # ------------------------------------------------------------
        # Per-file row-group progress bar
        # ------------------------------------------------------------
        row_group_bar = tqdm(
            total=n_row_groups,
            desc=f"Processing {file_name}",
            unit="group",
            leave=False
        )

        # Mark previously completed row-groups in the per-file progress bar
        row_group_bar.update(len(completed_groups))

        for rg_idx in range(n_row_groups):

            if rg_idx in completed_groups:
                continue

            table = pf.read_row_group(rg_idx, columns=[index_col, embedding_col])

            # Filter out rows where embedding is null
            non_null_mask = pc.is_valid(table[embedding_col])
            if non_null_mask.null_count == len(non_null_mask):
                completed_groups.add(rg_idx)
                state["completed"][file_name] = list(sorted(completed_groups))
                _atomic_write_json(state_file, state)
                row_group_bar.update(1)
                continue

            filtered = table.filter(non_null_mask)

            if filtered.num_rows == 0:
                completed_groups.add(rg_idx)
                state["completed"][file_name] = list(sorted(completed_groups))
                _atomic_write_json(state_file, state)
                row_group_bar.update(1)
                continue

            indices = filtered[index_col].to_pylist()

            emb_col = filtered[embedding_col]
            if isinstance(emb_col, pa.ChunkedArray):
                emb_values = []
                for c in emb_col.chunks:
                    emb_values.extend(c.to_pylist())
            else:
                emb_values = emb_col.to_pylist()

            for idx, emb in zip(indices, emb_values):
                idx_str = str(idx)

                image_indices.add(idx_str)

                if idx_str in current_image_indices:
                    valid_embedding_indices.add(idx_str)

            completed_groups.add(rg_idx)

            # Save progress
            state["completed"][file_name] = list(sorted(completed_groups))
            state["valid_embedding_indices"] = list(sorted(valid_embedding_indices))
            state["image_indices"] = list(sorted(image_indices))
            state["finished"] = False
            _atomic_write_json(state_file, state)

            row_group_bar.update(1)

        row_group_bar.close()  # ensure clean exit for next file

    # ------------------------------------------------------------
    # Final consistency check
    # ------------------------------------------------------------
    missing_images = valid_embedding_indices - image_indices
    extra_images = image_indices - valid_embedding_indices

    state["finished"] = True
    _atomic_write_json(state_file, state)

    return {
        "valid_embedding_count": len(valid_embedding_indices),
        "image_count": len(image_indices),
        "missing_images": missing_images,
        "extra_images": extra_images
    }

def _atomic_write_json(path, data):
    tmp = path + ".tmp"
    with open(tmp, "w") as f:
        json.dump(data, f, indent=2)
    os.replace(tmp, path)
```

```python
import os
import glob
import json
import pyarrow as pa
import pyarrow.parquet as pq
import pyarrow.compute as pc
from tqdm import tqdm


def reconcile_embeddings_and_images_safe(
        embeddings_dir,
        images_dir,
        state_file,
        index_col="image_index",
        embedding_col="embeddings_result",
        pattern="*.parquet"):

    # ------------------------------------------------------------
    # Load or initialize state
    # ------------------------------------------------------------
    if os.path.exists(state_file):
        with open(state_file, "r") as f:
            state = json.load(f)
    else:
        state = {
            "completed": {},  # {filename: [rowgroups]}
            "valid_embedding_indices": [],
            "finished": False
        }

    # Only embedding indices are tracked from state
    valid_embedding_indices = set(state["valid_embedding_indices"])

    # ------------------------------------------------------------
    # Scan images once (TRUE ground-truth image set)
    # ------------------------------------------------------------
    print(f"Scanning images in: {images_dir}")
    all_images = os.listdir(images_dir)
    image_indices = {os.path.splitext(f)[0] for f in all_images}  # THE REAL IMAGE SET

    print(f"Found {len(image_indices):,} images on disk")

    # ------------------------------------------------------------
    # Get Parquet files
    # ------------------------------------------------------------
    parquet_files = sorted(glob.glob(os.path.join(embeddings_dir, pattern)))
    print(f"Found {len(parquet_files)} embedding files")

    # ------------------------------------------------------------
    # Overall progress bar (files)
    # ------------------------------------------------------------
    for file_path in tqdm(parquet_files, desc="Overall Progress", unit="file"):

        file_name = os.path.basename(file_path)
        pf = pq.ParquetFile(file_path)
        n_row_groups = pf.num_row_groups

        completed_groups = set(state["completed"].get(file_name, []))

        # ------------------------------------------------------------
        # Per-file progress bar (row groups)
        # ------------------------------------------------------------
        row_group_bar = tqdm(
            total=n_row_groups,
            desc=f"Processing {file_name}",
            unit="group",
            leave=False
        )

        # Advance bar to already-done row groups
        row_group_bar.update(len(completed_groups))

        for rg_idx in range(n_row_groups):

            # Skip already processed
            if rg_idx in completed_groups:
                continue

            # Load this row group
            table = pf.read_row_group(rg_idx, columns=[index_col, embedding_col])

            # Filter null embeddings (fast in Arrow)
            non_null_mask = pc.is_valid(table[embedding_col])
            if non_null_mask.null_count == len(non_null_mask):
                # Entire group has no embeddings
                completed_groups.add(rg_idx)
                _save_state(state, state_file, file_name, completed_groups, valid_embedding_indices)
                row_group_bar.update(1)
                continue

            filtered = table.filter(non_null_mask)
            if filtered.num_rows == 0:
                completed_groups.add(rg_idx)
                _save_state(state, state_file, file_name, completed_groups, valid_embedding_indices)
                row_group_bar.update(1)
                continue

            # Extract indices and embeddings
            indices = filtered[index_col].to_pylist()

            emb_col = filtered[embedding_col]
            if isinstance(emb_col, pa.ChunkedArray):
                emb_values = []
                for c in emb_col.chunks:
                    emb_values.extend(c.to_pylist())
            else:
                emb_values = emb_col.to_pylist()

            # Validate: embedding is valid ONLY if the corresponding image exists
            for idx, emb in zip(indices, emb_values):
                idx_str = str(idx)
                if idx_str in image_indices:  # TRUE CHECK
                    valid_embedding_indices.add(idx_str)

            completed_groups.add(rg_idx)

            # Save state
            _save_state(state, state_file, file_name, completed_groups, valid_embedding_indices)

            row_group_bar.update(1)

        row_group_bar.close()

    # ------------------------------------------------------------
    # Final consistency check
    # ------------------------------------------------------------
    missing_images = valid_embedding_indices.symmetric_difference(image_indices) - image_indices
    extra_images = image_indices.symmetric_difference(valid_embedding_indices) - valid_embedding_indices

    # The above is equivalent to:
    # missing_images = valid_embedding_indices - image_indices
    # extra_images   = image_indices - valid_embedding_indices

    state["finished"] = True
    _atomic_write_json(state_file, state)

    return {
        "valid_embedding_count": len(valid_embedding_indices),
        "image_count": len(image_indices),
        "missing_images": valid_embedding_indices - image_indices,
        "extra_images": image_indices - valid_embedding_indices
    }


def _save_state(state, state_file, file_name, completed_groups, valid_embedding_indices):
    state["completed"][file_name] = list(sorted(completed_groups))
    state["valid_embedding_indices"] = list(sorted(valid_embedding_indices))
    state["finished"] = False
    _atomic_write_json(state_file, state)


def _atomic_write_json(path, data):
    tmp = path + ".tmp"
    with open(tmp, "w") as f:
        json.dump(data, f, indent=2)
    os.replace(tmp, path)
```
