## Model Evaluation Time

This section evaluates the inference execution speed of several object detection models on a 10,000-image subset of the LAION-5B dataset. The primary objective is to determine which models are practically applicable for large-scale analysis, balancing inference speed, model capacity, and dataset alignment.

All models were evaluated under consistent conditions, using an input resolution of 640 px unless otherwise stated.

### Inference Speed Summary

<div align="center">

| Model | Approx. Inference Time (10k images) | Notes |
|:-----:|:-----------------------------------:|:-----:|
| YOLOv8x | ~7 min 06 sec (~26 img/s) | High-capacity CNN detector |
| YOLOv8m | ~5 min 47 sec | Faster, lower-capacity variant |
| YOLOv8x-World | ~8 min 11 sec | Open-vocabulary, prompt-based |
| YOLOv8x-Worldv2 | ~7 min 23 sec | Improved open-vocabulary model |
| YOLOv8x + YOLOv8x-World (Ensemble) | ~12 min 44 sec | Near-additive inference cost |
| RT-DETR (RTDETR-X) | ~6 min 24 sec (~26 img/s) | Transformer-based real-time detector |
| GroundingDINO | > 1 hour | Computationally expensive |
| ViTDet | > 1 hour | Heavy transformer backbone |

</div>


From these results, GroundingDINO and ViTDet were excluded due to prohibitive inference times. The remaining models demonstrated sufficient throughput for large-scale experimentation.

### Model Selection Rationale

- **YOLOv8 family**  
    The YOLOv8 models represent a mature family of real-time, convolutional object detectors that are widely adopted in both research and industry. Pretrained variants are available directly on the OpenImagesV7 dataset, making them well-aligned with the evaluation task. Their strong performance, stable tooling, and reproducibility make them suitable baseline models for large-scale dataset analysis. As outlined in the [Ultralytics](https://docs.ultralytics.com/datasets/detect/open-images-v7/) page where the x & m variants reportadly achieved 36.3 & 33.6 map50-95 on the OpenImagev7 validation set.

- **YOLOv8-World**  
    YOLOv8-World extends the standard YOLOv8 architecture to an open-vocabulary setting, enabling object detection via textual prompts rather than a fixed class taxonomy. This capability allows the set of detectable objects to be expanded or refined without retraining, which is particularly useful in exploratory or bias-analysis scenarios. The model was included to assess whether prompt-based detection could offer flexibility benefits without incurring prohibitive computational cost. As outlined in the [Ultralytics](https://docs.ultralytics.com/models/yolo-world/#key-features) page where the x-World & x-Worldv2 variants reportadly achieved 47.0, 63.0, 51.2 & 36.3 & 47.1, 62.8, 51.4 mAP, mAP50, mAP75 on the COCO validation set.

- **Ensemble model**  
  A simple ensemble combining YOLOv8x and YOLOv8x-World was explored to assess whether complementary strengths could improve detection performance. However, inference time increased substantially, and evaluation complexity increased without clear empirical gains.

- **RT-DETR**  
    RT-DETR is a transformer-based real-time object detector designed to eliminate post-processing steps such as Non-Maximum Suppression. Despite its architectural complexity, RT-DETR demonstrated inference speeds comparable to YOLOv8x in this setup. It was included to evaluate whether a modern end-to-end transformer detector could provide competitive throughput and serve as a viable alternative to CNN-based detectors, however it has no pre-trained weights for the OpenImagesv7 dataset as its primarily trained on the COCO & Object365 dataset as per the [Ultralytics](https://docs.ultralytics.com/models/rtdetr/) page.

- **GroundingDINO**  
    GroundingDINO is an open-vocabulary vision–language model designed for phrase grounding and fine-grained object localization. It was selected for evaluation due to its reported state-of-the-art performance in zero-shot and open-vocabulary object detection benchmarks, including a 52.5 AP score on COCO zero-shot object detection (without COCO training data) and setting a new record on the ODinW zero-shot benchmark with a mean 26.1 AP as per the [paper](https://arxiv.org/pdf/2303.05499) making it a natural candidate for evaluating prompt-based detection on large, weakly curated datasets such as LAION-5B. However, its transformer-heavy architecture and cross-modal attention mechanisms resulted in inference times exceeding one hour for the 10k-image subset, rendering it impractical for large-scale deployment in this context.

- **ViTDet**  
    ViTDet is a Vision Transformer–based object detection framework that emphasizes high accuracy through large-scale transformer backbones. It was included to assess whether transformer-based detectors optimized for accuracy could be feasibly applied to large image collections. In practice, ViTDet exhibited prohibitively slow inference speeds on the evaluated subset, making it unsuitable for time-constrained or large-scale analyses despite its strong performance in benchmark settings. Model was trained on LVIS dataset.

### Yolo-World + Yolov8 Open Images

```python
import os
import torch
from ultralytics import YOLO
from tqdm import tqdm
from ensemble_boxes import weighted_boxes_fusion
import numpy as np
import cv2
import random

# --- Configuration ---
run_name = 'YOLOWorld_OIV7_Ensemble_WBF_Batched_2'
output_dir = r'C:\MastersRepos\LAION-5B Testing\YOLO_Ensemble_Results_Batched'
image_dir = r'C:\MastersRepos\ARI5902-Research-Topics-in-AI\LAION-5B Testing\Spurious_Feature\ImageCaptioningEvaluationDatasets\LAION-5B-10k\LAION-5B-10k-images'
device = 0 if torch.cuda.is_available() else 'cpu'
batch_size = 16

os.makedirs(output_dir, exist_ok=True)

# --- Load Models ---
model_world = YOLO('models/yolov8x-world.pt')
model_oiv7 = YOLO('models/yolov8x-oiv7.pt')
print("✅ Both YOLO models loaded (World + OIV7)")

# --- Get List of Images ---
image_files = [os.path.join(image_dir, f) for f in os.listdir(image_dir)
               if f.lower().endswith(('.jpg', '.png', '.jpeg'))]

# --- Set seed for reproducible colors ---
random.seed(42)

# --- Generate colors per class (for YOLOv8 OIV7) ---
all_class_names = model_oiv7.names  # dict: idx -> class_name
class_colors = {name: tuple(random.randint(0, 255) for _ in range(3)) 
                for name in all_class_names.values()}

# --- Helper: Normalize boxes for WBF ---
def normalize_boxes(boxes, img_shape):
    h, w = img_shape
    norm_boxes = boxes.copy()
    norm_boxes[:, [0, 2]] /= w
    norm_boxes[:, [1, 3]] /= h
    return norm_boxes

def denormalize_boxes(boxes, img_shape):
    h, w = img_shape
    boxes[:, [0, 2]] *= w
    boxes[:, [1, 3]] *= h
    return boxes

# --- Batched inference helper ---
def run_model_batch(model, batch_paths):
    results = model.predict(
        source=batch_paths,
        imgsz=640,
        device=device,
        verbose=False
    )
    batch_data = []
    for r, img_path in zip(results, batch_paths):
        boxes = r.boxes.xyxy.cpu().numpy()
        scores = r.boxes.conf.cpu().numpy()
        labels = r.boxes.cls.cpu().numpy()
        img = cv2.imread(img_path)
        h, w = img.shape[:2]
        norm_boxes = normalize_boxes(boxes, (h, w))
        batch_data.append((norm_boxes, scores, labels, (h, w)))
    return batch_data

# --- Batched Ensemble Inference ---
for i in tqdm(range(0, len(image_files), batch_size), desc="Processing Batches"):
    batch_paths = image_files[i:i+batch_size]

    # Run both models in batch
    world_batch = run_model_batch(model_world, batch_paths)
    oiv7_batch = run_model_batch(model_oiv7, batch_paths)

    # Fuse results per image
    for img_path, world_data, oiv7_data in zip(batch_paths, world_batch, oiv7_batch):
        boxes_a, scores_a, labels_a, (h, w) = world_data
        boxes_b, scores_b, labels_b, _ = oiv7_data

        if len(boxes_a) == 0 and len(boxes_b) == 0:
            continue

        # Weighted Boxes Fusion
        boxes_list = [boxes_a, boxes_b]
        scores_list = [scores_a, scores_b]
        labels_list = [labels_a, labels_b]

        boxes, scores, labels = weighted_boxes_fusion(
            boxes_list, scores_list, labels_list,
            weights=[1, 1],
        )

        # Convert back to pixel coordinates
        boxes = denormalize_boxes(boxes, (h, w))

        # --- Draw and Save Results ---
        img = cv2.imread(img_path)
        for box, score, label in zip(boxes, scores, labels):
            x1, y1, x2, y2 = map(int, box)
            class_name = all_class_names.get(int(label), str(int(label)))
            color = class_colors[class_name]

            # Draw bounding box
            cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)

            # Prepare label text
            label_text = f"{class_name} {score:.2f}"
            font = cv2.FONT_HERSHEY_SIMPLEX
            font_scale = 0.4
            thickness = 1

            # Get text size
            (text_width, text_height), baseline = cv2.getTextSize(label_text, font, font_scale, thickness)

            # Draw filled rectangle (background for text)
            cv2.rectangle(
                img,
                (x1, max(0, y1 - text_height - baseline)),
                (x1 + text_width, y1),
                color,
                thickness=-1
            )

            # Draw text in white
            cv2.putText(
                img,
                label_text,
                (x1, max(0, y1 - baseline)),
                font,
                font_scale,
                (255, 255, 255),
                thickness
            )

        save_path = os.path.join(output_dir, os.path.basename(img_path))
        cv2.imwrite(save_path, img)

print("✅ Batched ensemble inference complete. Results saved to:", output_dir)
```

```python
import os
import glob
import torch
import time
from PIL import Image
from typing import List, Dict, Any
from transformers import AutoProcessor, GroundingDinoForObjectDetection
from tqdm import tqdm

# ==========================================================
# 1. CONFIGURATION
# ==========================================================
MODEL_ID = "IDEA-Research/grounding-dino-tiny"
IMAGE_DIR = r"C:\MastersRepos\ARI5902-Research-Topics-in-AI\LAION-5B Testing\Spurious_Feature\ImageCaptioningEvaluationDatasets\LAION-5B-10k\LAION-5B-10k-images"

# Replace LVIS YAML with your Open Images classes
# --- Define custom text-based classes ---
open_images_classes = [
    "Accordion",
    "Adhesive tape",
    "Aircraft",
    "Airplane",
    "Alarm clock",
    "Alpaca",
    "Ambulance",
    "Animal",
    "Ant",
    "Antelope",
    "Apple",
    "Armadillo",
    "Artichoke",
    "Auto part",
    "Axe",
    "Backpack",
    "Bagel",
    "Baked goods",
    "Balance beam",
    "Ball",
    "Balloon",
    "Banana",
    "Band-aid",
    "Banjo",
    "Barge",
    "Barrel",
    "Baseball bat",
    "Baseball glove",
    "Bat (Animal)",
    "Bathroom accessory",
    "Bathroom cabinet",
    "Bathtub",
    "Beaker",
    "Bear",
    "Bed",
    "Bee",
    "Beehive",
    "Beer",
    "Beetle",
    "Bell pepper",
    "Belt",
    "Bench",
    "Bicycle",
    "Bicycle helmet",
    "Bicycle wheel",
    "Bidet",
    "Billboard",
    "Billiard table",
    "Binoculars",
    "Bird",
    "Blender",
    "Blue jay",
    "Boat",
    "Bomb",
    "Book",
    "Bookcase",
    "Boot",
    "Bottle",
    "Bottle opener",
    "Bow and arrow",
    "Bowl",
    "Bowling equipment",
    "Box",
    "Boy",
    "Brassiere",
    "Bread",
    "Briefcase",
    "Broccoli",
    "Bronze sculpture",
    "Brown bear",
    "Building",
    "Bull",
    "Burrito",
    "Bus",
    "Bust",
    "Butterfly",
    "Cabbage",
    "Cabinetry",
    "Cake",
    "Cake stand",
    "Calculator",
    "Camel",
    "Camera",
    "Can opener",
    "Canary",
    "Candle",
    "Candy",
    "Cannon",
    "Canoe",
    "Cantaloupe",
    "Car",
    "Carnivore",
    "Carrot",
    "Cart",
    "Cassette deck",
    "Castle",
    "Cat",
    "Cat furniture",
    "Caterpillar",
    "Cattle",
    "Ceiling fan",
    "Cello",
    "Centipede",
    "Chainsaw",
    "Chair",
    "Cheese",
    "Cheetah",
    "Chest of drawers",
    "Chicken",
    "Chime",
    "Chisel",
    "Chopsticks",
    "Christmas tree",
    "Clock",
    "Closet",
    "Clothing",
    "Coat",
    "Cocktail",
    "Cocktail shaker",
    "Coconut",
    "Coffee",
    "Coffee cup",
    "Coffee table",
    "Coffeemaker",
    "Coin",
    "Common fig",
    "Common sunflower",
    "Computer keyboard",
    "Computer monitor",
    "Computer mouse",
    "Container",
    "Convenience store",
    "Cookie",
    "Cooking spray",
    "Corded phone",
    "Cosmetics",
    "Couch",
    "Countertop",
    "Cowboy hat",
    "Crab",
    "Cream",
    "Cricket ball",
    "Crocodile",
    "Croissant",
    "Crown",
    "Crutch",
    "Cucumber",
    "Cupboard",
    "Curtain",
    "Cutting board",
    "Dagger",
    "Dairy Product",
    "Deer",
    "Desk",
    "Dessert",
    "Diaper",
    "Dice",
    "Digital clock",
    "Dinosaur",
    "Dishwasher",
    "Dog",
    "Dog bed",
    "Doll",
    "Dolphin",
    "Door",
    "Door handle",
    "Donut",
    "Dragonfly",
    "Drawer",
    "Dress",
    "Drill (Tool)",
    "Drink",
    "Drinking straw",
    "Drum",
    "Duck",
    "Dumbbell",
    "Eagle",
    "Earrings",
    "Egg (Food)",
    "Elephant",
    "Envelope",
    "Eraser",
    "Face powder",
    "Facial tissue holder",
    "Falcon",
    "Fashion accessory",
    "Fast food",
    "Fax",
    "Fedora",
    "Filing cabinet",
    "Fire hydrant",
    "Fireplace",
    "Fish",
    "Flag",
    "Flashlight",
    "Flower",
    "Flowerpot",
    "Flute",
    "Flying disc",
    "Food",
    "Food processor",
    "Football",
    "Football helmet",
    "Footwear",
    "Fork",
    "Fountain",
    "Fox",
    "French fries",
    "French horn",
    "Frog",
    "Fruit",
    "Frying pan",
    "Furniture",
    "Garden Asparagus",
    "Gas stove",
    "Giraffe",
    "Girl",
    "Glasses",
    "Glove",
    "Goat",
    "Goggles",
    "Goldfish",
    "Golf ball",
    "Golf cart",
    "Gondola",
    "Goose",
    "Grape",
    "Grapefruit",
    "Grinder",
    "Guacamole",
    "Guitar",
    "Hair dryer",
    "Hair spray",
    "Hamburger",
    "Hammer",
    "Hamster",
    "Hand dryer",
    "Handbag",
    "Handgun",
    "Harbor seal",
    "Harmonica",
    "Harp",
    "Harpsichord",
    "Hat",
    "Headphones",
    "Heater",
    "Hedgehog",
    "Helicopter",
    "Helmet",
    "High heels",
    "Hiking equipment",
    "Hippopotamus",
    "Home appliance",
    "Honeycomb",
    "Horizontal bar",
    "Horse",
    "Hot dog",
    "House",
    "Houseplant",
    "Human arm",
    "Human beard",
    "Human body",
    "Human ear",
    "Human eye",
    "Human face",
    "Human foot",
    "Human hair",
    "Human hand",
    "Human head",
    "Human leg",
    "Human mouth",
    "Human nose",
    "Humidifier",
    "Ice cream",
    "Indoor rower",
    "Infant bed",
    "Insect",
    "Invertebrate",
    "Ipod",
    "Isopod",
    "Jacket",
    "Jacuzzi",
    "Jaguar (Animal)",
    "Jeans",
    "Jellyfish",
    "Jet ski",
    "Jug",
    "Juice",
    "Kangaroo",
    "Kettle",
    "Kitchen & dining room table",
    "Kitchen appliance",
    "Kitchen knife",
    "Kitchen utensil",
    "Kitchenware",
    "Kite",
    "Knife",
    "Koala",
    "Ladder",
    "Ladle",
    "Ladybug",
    "Lamp",
    "Land vehicle",
    "Lantern",
    "Laptop",
    "Lavender (Plant)",
    "Lemon",
    "Leopard",
    "Light bulb",
    "Light switch",
    "Lighthouse",
    "Lily",
    "Limousine",
    "Lion",
    "Lipstick",
    "Lizard",
    "Lobster",
    "Loveseat",
    "Luggage and bags",
    "Lynx",
    "Magpie",
    "Mammal",
    "Man",
    "Mango",
    "Maple",
    "Maracas",
    "Marine invertebrates",
    "Marine mammal",
    "Measuring cup",
    "Mechanical fan",
    "Medical equipment",
    "Microphone",
    "Microwave oven",
    "Milk",
    "Miniskirt",
    "Mirror",
    "Missile",
    "Mixer",
    "Mixing bowl",
    "Mobile phone",
    "Monkey",
    "Moths and butterflies",
    "Motorcycle",
    "Mouse",
    "Muffin",
    "Mug",
    "Mule",
    "Mushroom",
    "Musical instrument",
    "Musical keyboard",
    "Nail (Construction)",
    "Necklace",
    "Nightstand",
    "Oboe",
    "Office building",
    "Office supplies",
    "Orange",
    "Organ (Musical Instrument)",
    "Ostrich",
    "Otter",
    "Oven",
    "Owl",
    "Oyster",
    "Paddle",
    "Palm tree",
    "Pancake",
    "Panda",
    "Paper cutter",
    "Paper towel",
    "Parachute",
    "Parking meter",
    "Parrot",
    "Pasta",
    "Pastry",
    "Peach",
    "Pear",
    "Pen",
    "Pencil case",
    "Pencil sharpener",
    "Penguin",
    "Perfume",
    "Person",
    "Personal care",
    "Personal flotation device",
    "Piano",
    "Picnic basket",
    "Picture frame",
    "Pig",
    "Pillow",
    "Pineapple",
    "Pitcher (Container)",
    "Pizza",
    "Pizza cutter",
    "Plant",
    "Plastic bag",
    "Plate",
    "Platter",
    "Plumbing fixture",
    "Polar bear",
    "Pomegranate",
    "Popcorn",
    "Porch",
    "Porcupine",
    "Poster",
    "Potato",
    "Power plugs and sockets",
    "Pressure cooker",
    "Pretzel",
    "Printer",
    "Pumpkin",
    "Punching bag",
    "Rabbit",
    "Raccoon",
    "Racket",
    "Radish",
    "Ratchet (Device)",
    "Raven",
    "Rays and skates",
    "Red panda",
    "Refrigerator",
    "Remote control",
    "Reptile",
    "Rhinoceros",
    "Rifle",
    "Ring binder",
    "Rocket",
    "Roller skates",
    "Rose",
    "Rugby ball",
    "Ruler",
    "Salad",
    "Salt and pepper shakers",
    "Sandal",
    "Sandwich",
    "Saucer",
    "Saxophone",
    "Scale",
    "Scarf",
    "Scissors",
    "Scoreboard",
    "Scorpion",
    "Screwdriver",
    "Sculpture",
    "Sea lion",
    "Sea turtle",
    "Seafood",
    "Seahorse",
    "Seat belt",
    "Segway",
    "Serving tray",
    "Sewing machine",
    "Shark",
    "Sheep",
    "Shelf",
    "Shellfish",
    "Shirt",
    "Shorts",
    "Shotgun",
    "Shower",
    "Shrimp",
    "Sink",
    "Skateboard",
    "Ski",
    "Skirt",
    "Skull",
    "Skunk",
    "Skyscraper",
    "Slow cooker",
    "Snack",
    "Snail",
    "Snake",
    "Snowboard",
    "Snowman",
    "Snowmobile",
    "Snowplow",
    "Soap dispenser",
    "Sock",
    "Sofa bed",
    "Sombrero",
    "Sparrow",
    "Spatula",
    "Spice rack",
    "Spider",
    "Spoon",
    "Sports equipment",
    "Sports uniform",
    "Squash (Plant)",
    "Squid",
    "Squirrel",
    "Stairs",
    "Stapler",
    "Starfish",
    "Stationary bicycle",
    "Stethoscope",
    "Stool",
    "Stop sign",
    "Strawberry",
    "Street light",
    "Stretcher",
    "Studio couch",
    "Submarine",
    "Submarine sandwich",
    "Suit",
    "Suitcase",
    "Sun hat",
    "Sunglasses",
    "Surfboard",
    "Sushi",
    "Swan",
    "Swim cap",
    "Swimming pool",
    "Swimwear",
    "Sword",
    "Syringe",
    "Table",
    "Table tennis racket",
    "Tablet computer",
    "Tableware",
    "Taco",
    "Tank",
    "Tap",
    "Tart",
    "Taxi",
    "Tea",
    "Teapot",
    "Teddy bear",
    "Telephone",
    "Television",
    "Tennis ball",
    "Tennis racket",
    "Tent",
    "Tiara",
    "Tick",
    "Tie",
    "Tiger",
    "Tin can",
    "Tire",
    "Toaster",
    "Toilet",
    "Toilet paper",
    "Tomato",
    "Tool",
    "Toothbrush",
    "Torch",
    "Tortoise",
    "Towel",
    "Tower",
    "Toy",
    "Traffic light",
    "Traffic sign",
    "Train",
    "Training bench",
    "Treadmill",
    "Tree",
    "Tree house",
    "Tripod",
    "Trombone",
    "Trousers",
    "Truck",
    "Trumpet",
    "Turkey",
    "Turtle",
    "Umbrella",
    "Unicycle",
    "Van",
    "Vase",
    "Vegetable",
    "Vehicle",
    "Vehicle registration plate",
    "Violin",
    "Volleyball (Ball)",
    "Waffle",
    "Waffle iron",
    "Wall clock",
    "Wardrobe",
    "Washing machine",
    "Waste container",
    "Watch",
    "Watercraft",
    "Watermelon",
    "Weapon",
    "Whale",
    "Wheel",
    "Wheelchair",
    "Whisk",
    "Whiteboard",
    "Willow",
    "Window",
    "Window blind",
    "Wine",
    "Wine glass",
    "Wine rack",
    "Winter melon",
    "Wok",
    "Woman",
    "Wood-burning stove",
    "Woodpecker",
    "Worm",
    "Wrench",
    "Zebra",
    "Zucchini"]


MAX_CLASSES_PER_CHUNK = 80
BOX_THRESHOLD = 0.25
TEXT_THRESHOLD = 0.5
BATCH_SIZE = 4

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# ==========================================================
# 2. HELPER FUNCTIONS
# ==========================================================
def structure_labels_for_grounding_dino(labels: List[str]) -> str:
    cleaned = [lbl.strip().strip(".").lower() for lbl in labels]
    prompt = ". ".join(cleaned)
    if not prompt.endswith("."):
        prompt += "."
    return prompt

def load_image(path: str) -> Image.Image:
    return Image.open(path).convert("RGB")

# ==========================================================
# 3. LOAD MODEL
# ==========================================================
processor = AutoProcessor.from_pretrained(MODEL_ID)
model = GroundingDinoForObjectDetection.from_pretrained(MODEL_ID).to(device)
model.eval()

# ==========================================================
# 4. PREPARE CLASS CHUNKS
# ==========================================================
num_classes = len(open_images_classes)
num_class_chunks = (num_classes + MAX_CLASSES_PER_CHUNK - 1) // MAX_CLASSES_PER_CHUNK
print(f"Loaded {num_classes} classes in {num_class_chunks} chunks.")

# ==========================================================
# 5. LOAD IMAGE PATHS
# ==========================================================
image_paths = sorted(glob.glob(os.path.join(IMAGE_DIR, "*.jpg")))
if not image_paths:
    raise FileNotFoundError(f"No images found in {IMAGE_DIR}")
print(f"Found {len(image_paths)} images.")

all_detections: List[List[Dict[str, Any]]] = [[] for _ in range(len(image_paths))]

# ==========================================================
# 6. BATCH INFERENCE WITH GLOBAL PROGRESS BAR
# ==========================================================
num_images = len(image_paths)

with tqdm(
    total=num_images,
    desc="Images processed",
    unit="img",
    dynamic_ncols=True
) as pbar:

    for batch_start in range(0, num_images, BATCH_SIZE):
        batch_paths = image_paths[batch_start:batch_start + BATCH_SIZE]
        batch_images = [load_image(p) for p in batch_paths]
        batch_target_sizes = [img.size[::-1] for img in batch_images]

        batch_start_time = time.time()

        for chunk_idx in range(num_class_chunks):
            start = chunk_idx * MAX_CLASSES_PER_CHUNK
            end = min((chunk_idx + 1) * MAX_CLASSES_PER_CHUNK, num_classes)
            class_chunk = open_images_classes[start:end]
            prompts = [structure_labels_for_grounding_dino(class_chunk)] * len(batch_images)

            inputs = processor(
                images=batch_images,
                text=prompts,
                return_tensors="pt"
            ).to(device)

            with torch.no_grad():
                with torch.autocast(
                    device_type=device,
                    dtype=torch.float16,
                    enabled=(device == "cuda")
                ):
                    outputs = model(**inputs)

            results = processor.post_process_grounded_object_detection(
                outputs,
                inputs.input_ids,
                threshold=TEXT_THRESHOLD,
                target_sizes=batch_target_sizes
            )

            for img_idx, res in enumerate(results):
                for score, label, box in zip(
                    res["scores"],
                    res["text_labels"],
                    res["boxes"]
                ):
                    if score >= BOX_THRESHOLD:
                        all_detections[batch_start + img_idx].append({
                            "score": score.item(),
                            "label": label.title(),
                            "box": [int(round(v.item())) for v in box],
                            "image_path": batch_paths[img_idx]
                        })

        batch_end_time = time.time()

        pbar.update(len(batch_paths))
        pbar.set_postfix({
            "batch_time_s": f"{batch_end_time - batch_start_time:.2f}",
            "batch_size": len(batch_paths)
        })

torch.cuda.empty_cache()
print("\n✅ Batched inference complete.")
```

```python
import os
os.environ["WANDB_DISABLED"] = "true"

from pathlib import Path
import json
import torch
import cv2
import numpy as np
from PIL import Image
from tqdm import tqdm

from detectron2.config import LazyConfig, instantiate
from detectron2.checkpoint import DetectionCheckpointer
from detectron2.data.transforms import ResizeShortestEdge
from detectron2.utils.visualizer import Visualizer, ColorMode

# ============================================================
# Configuration (mirrors YOLO template)
# ============================================================

run_name = "VitDet_Base_LVIS_Inference"

image_dir = r"C:\MastersRepos\ARI5902-Research-Topics-in-AI\LAION-5B Testing\Spurious_Feature\ImageCaptioningEvaluationDatasets\LAION-5B-10k\LAION-5B-10k-images"

output_dir = r"C:\MastersRepos\LAION-5B Testing\VitDet_LVIS_Detection_Results"

model_cfg = r"C:\MastersRepos\ARI5902-Research-Topics-in-AI\LAION-5B Testing\Spurious_Feature\ObjectDetection\detectron2\projects\ViTDet\configs\LVIS\cascade_mask_rcnn_vitdet_b_100ep.py"

checkpoint_path = r"C:\MastersRepos\ARI5902-Research-Topics-in-AI\LAION-5B Testing\Spurious_Feature\ObjectDetection\models\VitDet_base_LVIS.pkl"

color_json = r"C:\MastersRepos\ARI5902-Research-Topics-in-AI\LAION-5B Testing\Spurious_Feature\ObjectDetection\LVIS_color_map.json"

conf_thresh = 0.5
img_min_edge = 1024

device = "cuda" if torch.cuda.is_available() else "cpu"

# ============================================================
# Setup
# ============================================================

os.makedirs(output_dir, exist_ok=True)
run_dir = Path(output_dir) / run_name
run_dir.mkdir(parents=True, exist_ok=True)

# NumPy compatibility (Detectron2 + newer NumPy)
np.float_ = np.float = np.float64
np.complex_ = np.complex = np.complex128

# Load class colours
with open(color_json, "r") as f:
    color_map = json.load(f)

class_names = list(color_map.keys())
class_colors = list(color_map.values())

# ============================================================
# Load Model
# ============================================================

cfg = LazyConfig.load(model_cfg)
cfg.train.init_checkpoint = checkpoint_path

model = instantiate(cfg.model)
DetectionCheckpointer(model).load(cfg.train.init_checkpoint)
model.eval().to(device)

print(f"Model loaded: {Path(checkpoint_path).name} (LVIS)")

# ============================================================
# Image list (for tqdm parity with YOLO)
# ============================================================

image_files = [
    p for p in Path(image_dir).iterdir()
    if p.suffix.lower() in (".jpg", ".jpeg", ".png")
]

# ============================================================
# Inference Loop
# ============================================================

resize = ResizeShortestEdge(
    short_edge_length=img_min_edge,
    max_size=img_min_edge
)

total_detections = 0

with tqdm(total=len(image_files), desc="Processing Images", unit="img") as pbar:
    for img_path in image_files:
        img = Image.open(img_path).convert("RGB")
        img_np = np.array(img, dtype=np.uint8)

        tfm = resize.get_transform(img_np)
        resized = tfm.apply_image(img_np)

        img_tensor = torch.from_numpy(resized).permute(2, 0, 1).float().to(device)

        inputs = [{
            "image": img_tensor,
            "height": img_np.shape[0],
            "width": img_np.shape[1],
        }]

        with torch.no_grad(), torch.amp.autocast(device_type=device):
            outputs = model(inputs)

        instances = outputs[0]["instances"]
        keep = instances.scores > conf_thresh
        instances = instances[keep]

        num_det = len(instances)
        total_detections += num_det

        # ----------------------------------------------------
        # Visualization (YOLO-style save=True)
        # ----------------------------------------------------

        if num_det > 0:
            vis = Visualizer(
                img_np,
                instance_mode=ColorMode.IMAGE
            )

            for box, cls in zip(
                instances.pred_boxes.tensor.cpu().numpy(),
                instances.pred_classes.cpu().numpy()
            ):
                color = [c / 255 for c in class_colors[cls]]
                vis.draw_box(box, edge_color=color)
                vis.draw_text(class_names[cls], (box[0], box[1]), color=color)

            out_img = vis.get_output().get_image()
        else:
            out_img = img_np

        save_path = run_dir / img_path.name
        cv2.imwrite(str(save_path), cv2.cvtColor(out_img, cv2.COLOR_RGB2BGR))

        pbar.update(1)

print(f"✅ Total detections: {total_detections}")
```

## Model Selection for OpenImagesV7 Evaluation

While several detection models demonstrated adequate inference speed on a 10k-image subset of LAION-5B, only a subset of these models were suitable for rigorous evaluation on the OpenImagesV7 validation set. This section explains why the ensemble model and RT-DETR were excluded from OpenImagesV7 evaluation, and why the YOLOv8 and YOLOv8-World models were ultimately selected.

### Exclusion of the Ensemble Model

A simple ensemble combining YOLOv8x and YOLOv8x-World was initially explored to assess whether combining closed-set and open-vocabulary detectors could improve detection performance. However, this approach was not carried forward for OpenImagesV7 evaluation for the following reasons:

1. **Lack of Class-Space Consistency**  
   YOLOv8x is trained on a fixed OpenImagesV7 taxonomy, whereas YOLOv8-World operates in an open-vocabulary setting with prompt-defined classes. Aligning these two class spaces introduces ambiguity in class identity, label matching, and evaluation fairness, particularly for fine-grained OpenImages categories.

2. **Absence of Demonstrated Complementarity**  
   Preliminary validation indicated that YOLOv8x-OpenImages consistently outperformed YOLOv8-World across most classes. In such cases, ensembling a weaker model typically increases false positives and degrades precision rather than improving recall or mAP.

3. **Evaluation Complexity and Reproducibility**  
   Ensemble evaluation requires custom post-processing, class remapping, and fusion heuristics (e.g., weighted box fusion thresholds). These choices introduce additional degrees of freedom that complicate reproducibility and obscure attribution of performance gains.

4. **Suboptimal Performance of Open-Vocabulary Predictions**  
   Empirical evaluation showed that the YOLOv8-World models produced substantially weaker results on the OpenImagesV7 validation set compared to the closed-set YOLOv8x model. In particular, the open-vocabulary detectors exhibited lower precision and unstable recall across many OpenImages categories, reflecting limitations in prompt-based class specification for large, fine-grained taxonomies. Given these results, incorporating YOLOv8-World into an ensemble would likely degrade overall detection performance rather than improve it, making the ensemble approach unsuitable for reliable OpenImagesV7 evaluation.

Given these factors, the ensemble model was deemed unsuitable for a controlled, class-consistent evaluation on OpenImagesV7.

### Exclusion of RT-DETR

RT-DETR demonstrated inference speeds comparable to YOLOv8x on the 10k-image subset, motivating its initial inclusion. However, it was excluded from OpenImagesV7 evaluation for the following methodological reasons:

1. **Training Dataset Mismatch**  
   The RT-DETR models evaluated were pretrained Objects365. These datasets differ substantially from OpenImagesV7 in terms of class taxonomy, annotation density, and semantic granularity. Direct evaluation on OpenImagesV7 without retraining would therefore conflate dataset mismatch effects with model performance.

2. **Lack of Official OpenImagesV7 Benchmarks**  
   Unlike YOLOv8, RT-DETR does not provide published or documented performance metrics on OpenImagesV7. As a result, quantitative comparisons against OpenImages-trained detectors would lack a reliable baseline for interpretation.

3. **Evaluation Fairness**  
   Evaluating models trained on different label distributions without retraining or domain adaptation risks producing misleading conclusions. For a fair comparison, all evaluated models must share comparable training exposure to the target dataset.

Although RT-DETR is a strong real-time detector in its intended training regime, these constraints made it unsuitable for inclusion in the OpenImagesV7 evaluation presented in this study.

### Final Model Selection

Based on the above considerations, only YOLOv8 and YOLOv8-World were evaluated on the OpenImagesV7 validation set. Both models are directly compatible with the OpenImagesV7 taxonomy, and their evaluation can be conducted using standard, reproducible metrics.

Between these two, the standard YOLOv8-OpenImages model consistently achieved higher precision, recall, and mAP across the evaluated classes. Consequently, YOLOv8-OpenImages was selected as the primary detection model for subsequent analyses.

### Downloading Open Images v7 Dataset for evaluation

```python
import fiftyone as fo
import fiftyone.zoo as foz
from pathlib import Path
import yaml
import shutil
from ultralytics import YOLO
from collections import defaultdict
import json

def download_oiv7_validation(output_dir="oiv7_val", dataset_name="oiv7_val", max_samples=None):
    """
    Download Open Images v7 validation set with proper class index remapping for YOLOv8x-OIV7.
    """
    print("=" * 60)
    print("Downloading Open Images v7 Validation Dataset")
    print("=" * 60)

    # Load official mapping from YOLOv8x-OIV7
    model = YOLO("models/yolov8x-oiv7.pt")
    official_mapping = model.names  # Dict: {0: 'Accordion', 1: 'Adhesive tape', ...}
    
    # Create reverse mapping: name -> official_index
    name_to_official_idx = {name: idx for idx, name in official_mapping.items()}
    print(f"Official YOLOv8x-OIV7 mapping: {len(official_mapping)} classes")

    # Prepare output directory
    output_path = Path(output_dir).absolute()
    if output_path.exists():
        print(f"\nRemoving existing directory: {output_path}")
        shutil.rmtree(output_path)
        print("✓ Cleaned")

    # Step 1: Download dataset
    print("\nStep 1: Downloading dataset (this may take a while)...")
    dataset = foz.load_zoo_dataset(
        "open-images-v7",
        split="validation",
        label_types=["detections"],
        max_samples=max_samples,
        dataset_name=dataset_name
    )
    print(f"✓ Downloaded {len(dataset)} images")

    # Step 2: Detect the label field
    print("\nStep 2: Checking available fields...")
    detection_field = None
    for field_name in dataset.get_field_schema().keys():
        if "detection" in field_name.lower() or "ground_truth" in field_name.lower():
            detection_field = field_name
            print(f"✓ Using detection field: '{detection_field}'")
            break
    if detection_field is None:
        raise ValueError("No detection field found in dataset!")

    # Step 3: Export to YOLO format
    print("\nStep 3: Exporting to YOLO format...")
    dataset.export(
        export_dir=str(output_path),
        dataset_type=fo.types.YOLOv5Dataset,
        label_field=detection_field,
        split="val"
    )
    print(f"✓ Dataset exported to {output_path}")

    # Step 4: Read FiftyOne's class mapping
    print("\nStep 4: Reading FiftyOne's class mapping...")
    fiftyone_yaml_path = output_path / "dataset.yaml"
    
    with open(fiftyone_yaml_path, "r", encoding="utf-8") as f:
        fiftyone_yaml = yaml.safe_load(f)
    
    fiftyone_mapping = fiftyone_yaml.get("names", {})
    print(f"FiftyOne exported {len(fiftyone_mapping)} classes")
    
    # Create remapping: fiftyone_idx -> official_idx
    idx_remapping = {}
    unmapped_classes = []
    
    for fo_idx, class_name in fiftyone_mapping.items():
        if class_name in name_to_official_idx:
            official_idx = name_to_official_idx[class_name]
            idx_remapping[fo_idx] = official_idx
        else:
            unmapped_classes.append(class_name)
    
    print(f"✓ Mapped {len(idx_remapping)} classes")
    if unmapped_classes:
        print(f"⚠️  Warning: {len(unmapped_classes)} classes not in official mapping:")
        print(f"   {unmapped_classes[:10]}")

    # Step 5: Remap class indices in label files
    print("\nStep 5: Remapping class indices in label files...")
    labels_dir = output_path / "labels" / "val"
    label_files = list(labels_dir.glob("*.txt"))
    
    remapped_count = 0
    skipped_count = 0
    
    for label_file in label_files:
        new_lines = []
        modified = False
        
        with open(label_file, "r") as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) >= 5:
                    old_class_idx = int(parts[0])
                    
                    if old_class_idx in idx_remapping:
                        new_class_idx = idx_remapping[old_class_idx]
                        parts[0] = str(new_class_idx)
                        modified = True
                        new_lines.append(" ".join(parts))
                    else:
                        # Skip annotations for unmapped classes
                        skipped_count += 1
                        continue
                else:
                    new_lines.append(line.strip())
        
        if modified and new_lines:
            with open(label_file, "w") as f:
                f.write("\n".join(new_lines) + "\n")
            remapped_count += 1
    
    print(f"✓ Remapped {remapped_count} label files")
    if skipped_count > 0:
        print(f"⚠️  Skipped {skipped_count} annotations for unmapped classes")

    # Step 6: Create final dataset.yaml with official mapping
    print("\nStep 6: Creating dataset.yaml with official class mapping...")
    dataset_yaml = {
        "path": str(output_path),
        "train": "",
        "val": "images/val",
        "test": None,
        "names": official_mapping
    }

    with open(fiftyone_yaml_path, "w", encoding="utf-8") as f:
        yaml.dump(dataset_yaml, f, default_flow_style=False, sort_keys=False)

    print(f"✓ dataset.yaml updated with official 601-class mapping")
    
    # Step 7: Verification
    print("\nStep 7: Verifying remapping...")
    sample_label = list(label_files)[:1][0] if label_files else None
    if sample_label:
        with open(sample_label, "r") as f:
            lines = f.readlines()[:3]
            print(f"\nSample from {sample_label.name}:")
            for line in lines:
                class_idx = int(line.split()[0])
                class_name = official_mapping.get(class_idx, "UNKNOWN")
                print(f"  Class {class_idx}: {class_name}")

    print(f"\n{'='*60}")
    print("✅ Dataset ready for evaluation!")
    print(f"{'='*60}")
    print(f"Dataset structure:")
    print(f"  {output_path}/")
    print(f"  ├── images/val/")
    print(f"  ├── labels/val/ (✅ remapped to official indices)")
    print(f"  └── dataset.yaml (✅ official 601-class mapping)")

    return str(fiftyone_yaml_path)

# Example usage
output_dir = r"C:\MastersRepos\ARI5902-Research-Topics-in-AI\LAION-5B Testing\Spurious_Feature\ObjectDetection\OpenImagesv7_val_corrected"

# Download and remap
yaml_path = download_oiv7_validation(
    output_dir=output_dir, 
    dataset_name="oiv7_val_corrected", 
    max_samples=None
)

print(f"\n✅ Ready to evaluate!")
print(f"Dataset YAML: {yaml_path}")
```

### Evaluating Models

```python
import os
from ultralytics import YOLO

def evaluate_on_oiv7(model_path, dataset_yaml, classes=None, device=0, batch_size=16, save_dir="./runs/val"):
    """
    Evaluate a YOLO or YOLO-World model on Open Images v7 validation set.

    Args:
        model_path: Path to YOLO model weights
        dataset_yaml: Path to dataset.yaml file
        classes: List of class names for open-vocabulary YOLO-World (optional)
        device: GPU device index or 'cpu'
        batch_size: Batch size for evaluation
        save_dir: Directory to save evaluation results and plots
    """
    print("\n" + "="*60)
    print(f"Evaluating Model on Open Images v7: {model_path}")
    print("="*60)

    model = YOLO(model_path)
    
    # If classes are provided, set them (YOLO-World)
    if classes is not None:
        model.set_classes(classes)
        print(f"Classes set for open-vocabulary evaluation: {len(classes)} classes")

    # Make sure save directory exists
    os.makedirs(save_dir, exist_ok=True)

    # Run validation
    results = model.val(
        data=dataset_yaml,
        split='val',
        save_json=True,
        # conf=0.001,
        # iou=0.6,
        imgsz=640,
        max_det=300,
        plots=True,
        batch=batch_size,
        device=device
    )

    print("\n" + "="*60)
    print("Evaluation Results")
    print("="*60)
    print(f"mAP50: {results.box.map50:.4f}")
    print(f"mAP50-95: {results.box.map:.4f}")
    print(f"Precision: {results.box.mp:.4f}")
    print(f"Recall: {results.box.mr:.4f}")

    return results
```

## Hyperparameter Tuning

```python
import os
import time
import itertools
import random
import csv
import pandas as pd
import torch
from ultralytics import YOLO

def evaluate_on_oiv7(model, dataset_yaml, classes=None, device=0, batch_size=16, conf=None, iou=None, imgsz=640, augment=False, agnostic_nms=False):
    """
    Evaluate a YOLO or YOLO-World model on Open Images v7 validation set using a preloaded model.
    """
    print("\n" + "=" * 80)
    print(f"Evaluating Model on Open Images v7")
    print(f"Hyperparameters: conf={conf}, iou={iou}, imgsz={imgsz}, augment={augment}, agnostic_nms={agnostic_nms}")
    print("=" * 80)

    # if classes is not None:
    #     model.set_classes(classes)

    if classes is not None:
        try:
            # Move CLIP model to the same device
            if hasattr(model, 'model') and hasattr(model.model, 'clip_model'):
                model.model.clip_model = model.model.clip_model.to(device)

            # Move YOLO model to device
            model.to(device)

            # Set the classes safely
            model.set_classes(classes)

        except RuntimeError as e:
            if "Expected all tensors to be on the same device" in str(e):
                print("⚠️ CLIP text encoder device mismatch detected — retrying on CPU.")
                model.to("cpu")
                model.set_classes(classes)
            else:
                raise e

    start_time = time.time()
    results = model.val(
        data=dataset_yaml,
        split='val',
        save_json=False,
        conf=conf,
        iou=iou,
        imgsz=imgsz,
        max_det=300,
        augment=augment,
        agnostic_nms=agnostic_nms,
        plots=False,
        batch=batch_size,
        device=device,
        verbose=False
    )
    elapsed = time.time() - start_time

    print(f"Results: mAP50={results.box.map50:.4f}, mAP50-95={results.box.map:.4f}, "
          f"Precision={results.box.mp:.4f}, Recall={results.box.mr:.4f}, Time={elapsed:.1f}s")

    # Clear GPU cache
    torch.cuda.empty_cache()

    return results, elapsed

def tune_on_oiv7(model_path, dataset_yaml, classes=None, device=0, batch_size=16,
                 conf_values=None, iou_values=None, imgsz_values=None,
                 augment_values=None, agnostic_nms_values=None, search_samples=None,
                 save_dir="./runs/tuning_logs"):
    """
    Perform hyperparameter tuning for inference-time parameters to maximize mAP50-95.
    Resumable: skips already evaluated hyperparameter combinations if a CSV log exists.
    Loads the model only once and clears GPU cache after each run.
    """
    os.makedirs(save_dir, exist_ok=True)

    # Default search space
    if conf_values is None: conf_values = [0.001, 0.005, 0.01, 0.05]
    if iou_values is None: iou_values = [0.5, 0.6, 0.7]
    if imgsz_values is None: imgsz_values = [640, 800, 1024]
    if augment_values is None: augment_values = [True, False]
    if agnostic_nms_values is None: agnostic_nms_values = [True, False]

    param_grid = list(itertools.product(conf_values, iou_values, imgsz_values, augment_values, agnostic_nms_values))
    if search_samples is not None and search_samples < len(param_grid):
        param_grid = random.sample(param_grid, search_samples)

    log_path = os.path.join(save_dir, f"oiv7_tuning_results_{os.path.basename(model_path)}.csv")

    # Load existing log if present
    completed_combinations = set()
    best_map = 0.0
    best_params = {}
    if os.path.exists(log_path):
        df_existing = pd.read_csv(log_path)
        completed_combinations = set(tuple(x) for x in df_existing[['conf','iou','imgsz','augment','agnostic_nms']].values)
        if not df_existing.empty:
            idx_max = df_existing['mAP50_95'].idxmax()
            best_row = df_existing.loc[idx_max]
            best_map = best_row['mAP50_95']
            best_params = {
                'conf': best_row['conf'],
                'iou': best_row['iou'],
                'imgsz': best_row['imgsz'],
                'augment': best_row['augment'],
                'agnostic_nms': best_row['agnostic_nms']
            }

    # Load model once
    model = YOLO(model_path)

    # Open CSV in append mode
    with open(log_path, mode='a', newline='') as f:
        writer = csv.writer(f)
        if f.tell() == 0:
            writer.writerow(["conf", "iou", "imgsz", "augment", "agnostic_nms", "mAP50", "mAP50_95", "precision", "recall", "time_sec"])

        print(f"\n🔍 Starting hyperparameter tuning with {len(param_grid)} combinations...")
        print(f"Logging results to: {log_path}\n")

        best_results = None

        for idx, (conf, iou, imgsz, augment, agnostic_nms) in enumerate(param_grid, start=1):
            if (conf, iou, imgsz, augment, agnostic_nms) in completed_combinations:
                print(f"[{idx}/{len(param_grid)}] Skipping already evaluated: conf={conf}, iou={iou}, imgsz={imgsz}, augment={augment}, agnostic_nms={agnostic_nms}")
                continue

            print(f"\n[{idx}/{len(param_grid)}] Evaluating conf={conf}, iou={iou}, imgsz={imgsz}, augment={augment}, agnostic_nms={agnostic_nms}")
            results, elapsed = evaluate_on_oiv7(
                model=model,
                dataset_yaml=dataset_yaml,
                classes=classes,
                device=device,
                batch_size=batch_size,
                conf=conf,
                iou=iou,
                imgsz=imgsz,
                augment=augment,
                agnostic_nms=agnostic_nms
            )

            # Log results
            writer.writerow([
                conf, iou, imgsz, augment, agnostic_nms,
                results.box.map50, results.box.map,
                results.box.mp, results.box.mr,
                round(elapsed, 2)
            ])
            f.flush()

            # Track best
            current_map = results.box.map
            if current_map > best_map:
                best_map = current_map
                best_results = results
                best_params = {
                    'conf': conf,
                    'iou': iou,
                    'imgsz': imgsz,
                    'augment': augment,
                    'agnostic_nms': agnostic_nms
                }

    print("\n" + "=" * 80)
    print("🏆 Hyperparameter Tuning Complete")
    print("=" * 80)
    print(f"Best mAP50-95: {best_map:.4f}")
    print(f"Best Parameters: {best_params}")
    print(f"📘 Full results saved to: {log_path}")
    print("=" * 80)

    # Clear model from GPU
    model.cpu()
    del model
    torch.cuda.empty_cache()

    return best_results, best_params, log_path
```
