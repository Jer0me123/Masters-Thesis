## Filter Coco Datset

- This filters the Coco Dataset to only inlcude images depicting singular individuals.

```python
from pycocotools.coco import COCO
import shutil
from pathlib import Path
import os

root_path = Path('Coco2017Test')

coco = COCO(os.path.join(root_path, 'annotations/instances_val2017.json'))
person_cat_id = coco.getCatIds(catNms=['person'])[0]

output_dir = Path(os.path.join(root_path, 'val2017_single_person'))
output_dir.mkdir(exist_ok=True)

for img_id in coco.getImgIds():
    ann_ids = coco.getAnnIds(imgIds=img_id, catIds=person_cat_id)
    if len(ann_ids) == 1:  # Exactly one person
        img_info = coco.loadImgs(img_id)[0]
        src = Path(os.path.join(root_path, 'val2017')) / img_info['file_name']
        dst = output_dir / img_info['file_name']
        shutil.copy(src, dst)

print(f"Copied {len(list(output_dir.glob('*.jpg')))} single-person images")
```

### Updating Yolact Code

- These are minimal code updates which had to be applied for the code to function.

```python
from pathlib import Path

yolact_file = Path("yolact/yolact.py")

old = 'self.priors = torch.Tensor(prior_data, device=device).view(-1, 4).detach()'
new = 'self.priors = torch.tensor(prior_data, device=device).view(-1, 4).detach()'

text = yolact_file.read_text(encoding="utf-8")

if old not in text:
    raise RuntimeError("Target line not found — file may differ from expected version.")

yolact_file.write_text(text.replace(old, new), encoding="utf-8")

print("Update applied successfully.")
```

```python
from pathlib import Path
import re

file_path = Path("yolact/layers/output_utils.py")

text = file_path.read_text(encoding="utf-8")

pattern = re.compile(
    r"(?P<indent>\s*)# Test flag, do not upvote\s*\n"
    r"(?P=indent)if cfg\.mask_proto_debug:\s*\n"
    r"(?P<indent2>\s*)np\.save\('scripts/proto\.npy', proto_data\.cpu\(\)\.numpy\(\)\)"
)

def replacer(match):
    indent = match.group("indent")
    indent2 = match.group("indent2")
    return (
        f"{indent}# # Test flag, do not upvote\n"
        f"{indent}# if cfg.mask_proto_debug:\n"
        f"{indent2}#     np.save('scripts/proto.npy', proto_data.cpu().numpy())"
    )

new_text, count = pattern.subn(replacer, text)

if count != 1:
    raise RuntimeError(f"Expected 1 replacement, got {count}")

file_path.write_text(new_text, encoding="utf-8")

print("Block successfully commented out.")
```
